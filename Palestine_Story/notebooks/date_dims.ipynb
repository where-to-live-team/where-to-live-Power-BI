```python
d = pd.DataFrame({'date': pd.date_range('2007-01-01', '2026-12-31')})
d['date_key']        = d['date'].dt.strftime('%Y%m%d').astype(int)
d['year']            = d['date'].dt.year
d['quarter']         = d['date'].dt.quarter
d['month']           = d['date'].dt.month
d['month_name']      = d['date'].dt.month_name()
d['year_month']      = d['date'].dt.strftime('%Y-%m')
d['month_start']     = d['date'].dt.to_period('M').dt.start_time.dt.date
d['week_start']      = (d['date'] - pd.to_timedelta(d['date'].dt.weekday, unit='D')).dt.date
d['is_story_period'] = d['year'].between(2016, 2026).astype(int)
d['is_war_period']   = (d['date'] >= '2023-10-07').astype(int)
d['date']            = d['date'].dt.date
d = d[['date_key','date','year','quarter','month','month_name','year_month',
       'month_start','week_start','is_story_period','is_war_period']]

y = pd.DataFrame({'year': range(1960, 2027)})
y['is_story_period'] = y['year'].between(2016, 2026).astype(int)
y['is_war_year']     = (y['year'] >= 2023).astype(int)

assert d['date_key'].is_unique and len(d) == 7305
save_clean(d, 'dim_date.csv'); save_clean(y, 'dim_year.csv')
```
